## Step 2 – Create the complete project automatically
Run this cell once. It creates all required Python files, pages, styling, dependencies, and README.

````python
from pathlib import Path

PROJECT_DIR = Path.cwd() / "data_science_portfolio"
PROJECT_DIR.mkdir(exist_ok=True)

PROJECT_FILES = {}

for relative_path, content in PROJECT_FILES.items():
    target = PROJECT_DIR / relative_path
    target.parent.mkdir(parents=True, exist_ok=True)
    target.write_text(content, encoding="utf-8")

print(f"Created {len(PROJECT_FILES)} files in: {PROJECT_DIR.resolve()}")
print("Main application:", PROJECT_DIR / "app.py")
````

## Step 3 – Check that the Python files have no syntax errors

This checks every `.py` file before you launch the website.

````python
import ast

python_files = list(PROJECT_DIR.rglob("*.py"))
for path in python_files:
    ast.parse(path.read_text(encoding="utf-8"), filename=str(path))

print(f"Syntax check passed for {len(python_files)} Python files.")
````

## Step 4 – Launch the Streamlit website

Run this cell after installation and project creation. It starts Streamlit in the background and prints the local URL.

If the process is already running, do not start another copy. If you change the app code, stop the old process and run this cell again.

````python
import subprocess
import sys
import time
import urllib.request

# Stop a previously launched process from this notebook, if one exists.
try:
    if "streamlit_process" in globals() and streamlit_process.poll() is None:
        streamlit_process.terminate()
        streamlit_process.wait(timeout=5)
except Exception:
    pass

streamlit_process = subprocess.Popen(
    [
        sys.executable, "-m", "streamlit", "run",
        str(PROJECT_DIR / "app.py"),
        "--server.headless", "true",
        "--server.port", "8501",
    ],
    cwd=str(PROJECT_DIR),
    stdout=subprocess.DEVNULL,
    stderr=subprocess.STDOUT,
)

time.sleep(4)
print("Streamlit process running:", streamlit_process.poll() is None)
print("Open your portfolio at: http://localhost:8501")
````
