## 3. Generate the project files with error handling

````python
created_files = []
write_errors = []

try:
    PROJECT_DIR.mkdir(parents=True, exist_ok=True)
except OSError as exc:
    raise RuntimeError(f"Could not create project directory {PROJECT_DIR}: {exc}") from exc

for relative_path, content in PROJECT_FILES.items():
    target_path = PROJECT_DIR / relative_path
    try:
        target_path.parent.mkdir(parents=True, exist_ok=True)
        target_path.write_text(content, encoding="utf-8")
        created_files.append(relative_path)
    except OSError as exc:
        write_errors.append((relative_path, str(exc)))
        print(f"ERROR writing {relative_path}: {exc}")

if write_errors:
    failed_names = ", ".join(name for name, _ in write_errors)
    raise RuntimeError(f"Project generation failed for: {failed_names}")

required_files = [
    "app.py", "requirements.txt", "README.md",
    "pages/1_Projects.py", "pages/2_Interactive_ML_Demo.py",
    "pages/3_Data_Explorer.py", "pages/4_About_Me.py",
    "utils/data.py", "utils/ui.py", "assets/style.css",
]
missing = [name for name in required_files if not (PROJECT_DIR / name).is_file()]
if missing:
    raise FileNotFoundError(f"Required project files are missing: {missing}")

print(f"SUCCESS: Generated {len(created_files)} files in {PROJECT_DIR.resolve()}")
````

## 4. Validate Python syntax

````python
import ast

python_files = sorted(PROJECT_DIR.rglob("*.py"))
if not python_files:
    raise FileNotFoundError(
        f"No Python files found in {PROJECT_DIR}. Run the file-generation cell first."
    )

syntax_errors = []
for python_file in python_files:
    try:
        ast.parse(python_file.read_text(encoding="utf-8"), filename=str(python_file))
    except (SyntaxError, OSError) as exc:
        syntax_errors.append((str(python_file.relative_to(PROJECT_DIR)), str(exc)))

if syntax_errors:
    for filename, error in syntax_errors:
        print(f"FAILED: {filename}: {error}")
    raise SyntaxError(f"{len(syntax_errors)} Python file(s) failed syntax validation.")

print(f"SUCCESS: All {len(python_files)} Python files passed syntax validation.")
````

## 5. Install the generated project's requirements

````python
import subprocess
import sys

result = subprocess.run(
    [sys.executable, "-m", "pip", "install", "-r", str(PROJECT_DIR / "requirements.txt")],
    capture_output=True,
    text=True,
)
print(result.stdout[-3000:])
if result.returncode != 0:
    print(result.stderr[-3000:])
    raise RuntimeError("Dependency installation failed. Review the output above.")

print("SUCCESS: Dependencies installed.")
````

## 6. Run the Streamlit website

The following cell launches Streamlit on port 8501. Open **http://localhost:8501** in your browser.

````python
import subprocess
import time
import sys

try:
    if "streamlit_process" in globals() and streamlit_process.poll() is None:
        streamlit_process.terminate()
        streamlit_process.wait(timeout=5)
except Exception as exc:
    print(f"Previous process note: {exc}")

streamlit_process = subprocess.Popen(
    [
        sys.executable, "-m", "streamlit", "run", str(PROJECT_DIR / "app.py"),
        "--server.headless", "true", "--server.port", "8501",
    ],
    cwd=str(PROJECT_DIR),
    stdout=subprocess.DEVNULL,
    stderr=subprocess.STDOUT,
)

time.sleep(5)
if streamlit_process.poll() is not None:
    raise RuntimeError("Streamlit stopped during startup. Check dependencies and terminal output.")

print("Streamlit is running.")
print("Open: http://localhost:8501")
print("To stop it, run: streamlit_process.terminate()")
````
